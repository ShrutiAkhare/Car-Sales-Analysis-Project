```python
text_columns = [
    'name',
    'fuel',
    'seller_type',
    'transmission',
    'owner'
]

for col in text_columns:
    df_clean[col] = df_clean[col].str.strip()
```

```python
# Check unique values:
for col in ['fuel', 'seller_type', 'transmission', 'owner']:
    print(f"\n{col}:")
    print(df_clean[col].unique())
```

```python
# Extract mileage number
# \d+: Matches one or more digits (0–9). For example, in "18.5 kmpl", this grabs "18".
# \.?: Matches zero or one literal decimal point. The backslash \ treats the dot as an actual period rather than a wildcard, and the ? makes it optional so whole numbers like "1200 CC" still match.
# \d*: Matches zero or more digits after the decimal point. In "18.5 kmpl", this grabs "5".

df_clean['mileage_value'] = pd.to_numeric(
    df_clean['mileage'].str.extract(r'(\d+\.?\d*)')[0],
    errors='coerce'
)

# Extract engine capacity
df_clean['engine_cc'] = pd.to_numeric(
    df_clean['engine'].str.extract(r'(\d+\.?\d*)')[0],
    errors='coerce'
)

# Extract maximum power
df_clean['max_power_bhp'] = pd.to_numeric(
    df_clean['max_power'].str.extract(r'(\d+\.?\d*)')[0],
    errors='coerce'
)
```
